```python
import pandas as pd
```

```python
reviews_df = pd.read_csv("../data_preparation/glassdoor_reviews_annotated.csv")
```

```python
reviews_df.head()
```

```python
reviews_df["annotated"] = 0
```

```python
reviews_df["sentiment"].value_counts()
```

```python
reviews_df["sentiment"] = reviews_df["sentiment"].apply(
    lambda x: 2 if x == -1 else x
)
```

```python
reviews_df["sentiment"].value_counts()
```

```python
num_samples = int(20*240/100)
num_samples
```

```python
random_positive_reviews_df = reviews_df[reviews_df["sentiment"] == 1].sample(
    n=num_samples
)
```

```python
random_positive_reviews_df["sentiment"].value_counts()
```

```python
random_negative_reviews_df = reviews_df[reviews_df["sentiment"] == 2].sample(
    n=num_samples
)
```

```python
random_negative_reviews_df["sentiment"].value_counts()
```

```python
random_neutral_reviews_df = reviews_df[reviews_df["sentiment"] == 0].sample(
    n=num_samples
)
```

```python
random_neutral_reviews_df["sentiment"].value_counts()
```

```python
golden_df = pd.concat(
    [
        random_positive_reviews_df,
        random_negative_reviews_df,
        random_neutral_reviews_df,
    ]
)
```

```python
golden_df["sentiment"].value_counts()
```

```python
golden_df.head(5)
```

```python
golden_df.to_csv("golden_reviews_df.csv", index=False)
```
